```python
#Q6.Create a function to generate a fibonacci series
def generate_fibonacci(n):
    if n <= 0:
        return []
    elif n == 1:
        return [0]
    fib_series = [0, 1]
    while len(fib_series) < n:
        next_term = fib_series[-1] + fib_series[-2]
        fib_series.append(next_term)

    return fib_series
print(generate_fibonacci(10))
```

```python
#Q7.`Create using functions that contains various features
def add(num1,num2):
  return num1+num2

def sub(num1,num2):
  return num1-num2

def mul(num1,num2):
  return num1*num2

def div(num1,num2):
  if num2==0:
    return "zero division error"
  else:
    return num1-num2

num1=int(input("enter the number 1:"))
num2=int(input("enter the number 2:"))
print("enter the operator '+' for addition")
print("enter the operator '-' for subtraction")
print("enter the operator '*'for multiplication")
print("enter the operator '/'for division")
operator =input("enter the operator:")
if operator =="+":
  print(add(num1,num2))
elif  operator =="-":
  print(sub(num1,num2))
elif operator =="*":
  print(mul(num1,num2))
elif operator =="/":
  print(div(num1,num2))
```

```python
#Q8.Create a text file and store studetn details .
def save_student_details():
    name = input("Enter student name: ")
    roll_no = input("Enter roll number: ")
    grade = input("Enter the grade:")
    with open("students.txt", "a") as file:
        record = f"Name: {name}, Roll No: {roll_no}, Grade: {grade}\n"
        file.write(record)
if __name__ == "__main__":
    save_student_details()

```

```python
#Q9.Read data from file
def read_student_details():
    try:
        with open("students.txt", "r") as file:
            content = file.read()
            if not content.strip():
                print("The file 'students.txt' is empty.")
            else:
                print("--- Student Records ---")
                print(content)

    except FileNotFoundError:
        print("Error: 'students.txt' not found. Please add student details first.")
if __name__ == "__main__":
    read_student_details()
```
